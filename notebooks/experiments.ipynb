# Paving segmentation experiments

This notebook evaluates the application pipeline on a local image. Install the project with `pip install -e .` from the repository root and set `HF_TOKEN` before running it.

```python
from pathlib import Path
from PIL import Image
from paving_measurement.config import load_settings
from paving_measurement.segmentation import Sam3Segmenter

settings = load_settings()
segmenter = Sam3Segmenter(settings)
```

```python
IMAGE_PATH = Path('data/example.jpg')  # Replace with a local satellite image.
LATITUDE = 35.996421
ZOOM = 18

image = Image.open(IMAGE_PATH).convert('RGB')
result = segmenter.run(image, latitude=LATITUDE, zoom=ZOOM)
result.final_overlay
```

```python
print(result.summary)
result.comparison
```
